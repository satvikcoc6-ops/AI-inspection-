# Try the trained defect model
This notebook runs inference, not training. Upload the supplied ZIP when prompted. This is a metal-nut anomaly baseline, not a general automotive inspector. The included scratch example is a documented missed detection.

```python
from google.colab import files
import zipfile, pathlib
uploaded = files.upload()
package = next(name for name in uploaded if name.endswith(".zip"))
with zipfile.ZipFile(package) as archive:
    archive.extractall("/content/mdm")
project = pathlib.Path("/content/mdm/mdm_defect_model")
print("Package extracted")
```

```python
import subprocess, sys
from IPython.display import display, Image
subprocess.run([sys.executable, str(project/"predict.py"), str(project/"examples/scratch.png"), "--output", "/content/result"], check=True)
display(Image(filename="/content/result/heatmap_overlay.png"))
print(pathlib.Path("/content/result/result.json").read_text())
```
